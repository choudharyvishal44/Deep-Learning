```python
import tensorflow 
from tensorflow import keras
from keras.models import load_model
```

```python
model = load_model(r"D:\DeepLearning.py\DeepLearningProjects\DogVsCatClassifier\DogVsCatClassifier.keras")
```

```python
import cv2
import numpy as np
import os


folder_path = r"D:\DeepLearning.py\DeepLearningProjects\DogVsCatClassifier\test\cats"
classification = []
i = 0
for image_name in os.listdir(folder_path):

    image_path = os.path.join(folder_path, image_name)

    img = cv2.imread(image_path)

    if img is None:
        continue

    img = cv2.resize(img, (256, 256))
    img = img / 255.0
    img = np.expand_dims(img, axis=0)

    prediction = model.predict(img, verbose=0)

    if prediction[0][0] >= 0.5:
        result = "Dog"
    else:
        result = "Cat"

    classification.append(result)
    print(image_name, "->", result)
    if 199<= i <= 201:
        break
    i+=1
```

```python
clf = np.array(classification)
```
