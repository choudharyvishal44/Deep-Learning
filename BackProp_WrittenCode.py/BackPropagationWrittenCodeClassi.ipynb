# BackPropagation Written Code (Classification)

```python
import numpy as np
import pandas as pd
```

```python
df = pd.DataFrame([[8,8,1],[7,9,1],[6,10,0],[5,5,0]], columns=['cgpa', 'profile_score', 'placed'])
```

```python
def initialize_parameters(layer_dims):
    np.random.seed(3)
    parameters = {}
    L = len(layer_dims)

    for i in range(1,L):
        parameters["W" + str(i)] = np.ones((layer_dims[i-1], layer_dims[i])) * 0.1
        parameters['b' + str(i)] = np.zeros((layer_dims[i], 1))

    return parameters
```

```python
# Utility Function

def sigmoid(Z):
    A = 1/(1+np.exp(-Z))
    return A
```

```python
def linear_forward(A_prev, W,b):
    Z = np.dot(W.T, A_prev) + b
    A = sigmoid(Z)
    return A
```

```python
def L_layer_forward(X,parameters,layer_dims):
    A = X
    L = len(layer_dims)

    for i in range(1,L):
        A_prev = A
        A = linear_forward(A_prev,parameters["W" + str(i)],parameters["b" + str(i)])

    return A, A_prev
```

```python
def update_parameters(A,A_prev,parameters,X,y):
    y_hat = A
    lr = 0.001

    parameters["W2"][0][0] = parameters['W2'][0][0] + lr * (y-y_hat) * A_prev[0][0]
    parameters['W2'][1][0] = parameters['W2'][1][0] + lr * (y-y_hat) * A_prev[1][0]
    parameters['b2'] = parameters['b2'] + lr * (y-y_hat)

    parameters['W1'][0][0] = parameters['W1'][0][0] + lr * (y-y_hat) * parameters['W2'][0][0] * A_prev[0][0] * (1 - A_prev[0][0]) * X[0][0]
    parameters['W1'][1][0] = parameters['W1'][1][0] + lr * (y-y_hat) * parameters['W2'][0][0] * A_prev[0][0] * (1 - A_prev[0][0]) * X[1][0]
    parameters['b1'][0][0] = parameters['b1'][0][0] + lr * (y-y_hat) * parameters['W2'][0][0] * A_prev[0][0] * (1 - A_prev[0][0])

    parameters['W1'][0][1] = parameters['W1'][0][1] + lr * (y-y_hat) * parameters['W2'][1][0] * A_prev[1][0] * (1 - A_prev[1][0]) * X[0][0]
    parameters['W1'][1][1] = parameters['W1'][1][1] + lr * (y-y_hat) * parameters['W2'][1][0] * A_prev[1][0] * (1 - A_prev[1][0]) * X[1][0]
    parameters['b1'][1][0] = parameters['b1'][1][0] + lr * (y-y_hat) * parameters['W2'][1][0] * A_prev[1][0]* (1 - A_prev[1][0]) 
```

```python
epochs = 75
X = df.drop(columns = ['placed'])
y = df[['placed']].values
parameters = initialize_parameters([2,2,1])
np.random.seed(7)

for i in range(1, epochs + 1):
    losses = []
    for j in range(X.shape[0]):
        input = (X.iloc[j]).values.reshape(2,1)
        original = y[j][0]
        A, A_prev = L_layer_forward(input, parameters, [2,2,1])
        A = A[0][0]
        update_parameters(A,A_prev,parameters,input,original)
        loss = -original*np.log(A) - (1-original) * np.log(1-A)
        losses.append(loss)
    print(f'Epoch: {i} -> Loss = {np.array(losses).mean()}')
```
