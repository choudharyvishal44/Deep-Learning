# Backpropagation Written Code (Regression)

```python
import numpy as np
import pandas as pd
```

```python
df = pd.DataFrame([[8,8,4],[7,9,5],[6,10,6],[5,12,7]], columns= ['cpga','profile_score','lpa'])
```

```python
df
```

```python
X = df.drop(columns = ['lpa'])
y = df['lpa']
```

```python
def initialize_parameters(layer_dims):
    np.random.seed(3)
    parameters = {}
    L = len(layer_dims)

    for i in range(1,L):
        parameters['W' + str(i)] = np.ones((layer_dims[i - 1], layer_dims[i])) * 0.1
        parameters['b' + str(i)] = np.zeros((layer_dims[i],1))

    return parameters
```

```python
def linear_forward(A_prev, W,b):
    return np.dot(W.T, A_prev) + b
```

```python
# Forward Propagation
def L_layer_forward(X,parameters,layer_dims):
    A = X
    L = len(layer_dims)

    for i in range(1,L):
        A_prev = A
        A = linear_forward(A_prev,parameters['W' + str(i)], parameters['b' + str(i)])


    return A[0], A_prev
```

```python
def update_parameters(parameters,A,A_prev,X,y):
    y_hat = A[0]
    loss = (y-y_hat)
    lr = 0.001
    parameters['W1'][0][0] = parameters['W1'][0][0] + lr * 2 * loss * parameters['W2'][0][0] * X[0][0]
    parameters['W1'][1][0] = parameters['W1'][1][0] + lr * 2 * loss * parameters['W2'][0][0] * X[1][0]
    parameters['b1'][0][0] = parameters['b1'][0][0] + lr * 2 * loss * parameters['W2'][0][0]

    parameters['W1'][0][1] = parameters['W1'][0][1] + lr * 2 * loss * parameters['W2'][1][0] * X[0][0]
    parameters['W1'][1][1] = parameters['W1'][1][1] + lr * 2 * loss * parameters['W2'][1][0] * X[1][0]
    parameters['b1'][1][0] = parameters['b1'][1][0] + lr * 2 * loss * parameters['W2'][1][0]

    parameters['W2'][0][0] = parameters['W2'][0][0] + lr * 2 * loss * A_prev[0][0]
    parameters['W2'][1][0] = parameters['W2'][1][0] + lr * 2 * loss * A_prev[1][0]
    parameters['b2'] = parameters['b2'] + lr * 2 * loss  

    return parameters, loss**2
```

```python
# epochs implementation
parameters = initialize_parameters([2,2,1])
epochs = 20

for i in range(1,epochs+1):
    losses = []
    np.random.seed(7)
    for k in range(X.shape[0]):
        j = np.random.randint(X.shape[0])
        input  = np.array(X.iloc[j]).reshape(2,1)
        original = y[j]
        A, A_prev = L_layer_forward(input ,parameters,[2,2,1])
        parameters,loss  = update_parameters(parameters,A,A_prev, input,original)
        losses.append(loss)
    print(f"Epoch: {i} -> Loss : {np.array(losses).mean()}")


    
```
