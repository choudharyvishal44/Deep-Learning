# BackPropagation Using Keras For Regression

```python
import numpy as np
import pandas as pd
```

```python
df = pd.DataFrame([[8,8,4],[7,9,5],[6,10,6],[5,12,7]], columns= ['cpga','profile_score','lpa'])
```

```python
df
```

```python
X = np.array(df.drop(columns = ['lpa']))
y =  df['lpa']
```

```python
import tensorflow 
from tensorflow import keras 
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Dense
```

```python
model = Sequential()

model.add(Dense(2, activation = 'linear',input_dim = 2))
model.add(Dense(1, activation = 'linear'))
```

```python
model.summary()
```

```python
model.get_weights()
```

```python
new_weights = [np.array([[0.1,0.1],[0.1,0.1]],dtype = np.float32),np.array([0,0], dtype = np.float32),np.array([[0.1],[0.1]],dtype = np.float32),np.array([0],dtype = np.float32)]
```

```python
model.set_weights(new_weights)
```

```python
optimizer = keras.optimizers.Adam(learning_rate= 0.001)
model.compile(loss = 'mean_squared_error', optimizer = optimizer)
```

```python
model.fit(df.iloc[:,0:-1].values, df['lpa'].values,epochs = 75, verbose = 1, batch_size= 1)
```

```python
model.get_weights()
```

# BackPropagation Using Keras For Classification

```python
model2 = Sequential()

model2.add(Dense(2,activation = 'sigmoid', input_dim =2))
model2.add(Dense(1,activation='sigmoid'))
```

```python
df = pd.DataFrame([[8,8,1],[7,9,1],[6,10,0],[5,5,0]], columns=['cgpa', 'profile_score', 'placed'])
```

```python
model2.summary()
```

```python
model2.get_weights()
```

```python
new_weights = [np.array([[0.1,0.1],[0.1,0.1]],dtype = np.float32),np.array([0,0],dtype = np.float32),np.array([[0.1],[0.1]], dtype = np.float32), np.array([0], dtype = np.float32)]
```

```python
model2.set_weights(new_weights)
```

```python
optimizer = keras.optimizers.Adam(learning_rate= 0.001)
model2.compile(loss = 'binary_crossentropy', optimizer = optimizer)
```

```python
model2.fit(df.drop(columns = ['placed']).values, df['placed'].values, epochs = 75, verbose = 1, batch_size= 1)
```
