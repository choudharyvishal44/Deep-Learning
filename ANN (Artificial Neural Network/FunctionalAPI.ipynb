# **Simple Functional API**

```python
import tensorflow
from tensorflow import keras
from keras.models import Model
from keras.layers import *
```

```python
input = Input(shape = (3,))
```

```python
hidden1 = Dense(128, activation = 'relu')(input)
hidden2 = Dense(64, activation = 'relu')(hidden1)

output1 = Dense(1, activation = 'linear')(hidden2)
output2 = Dense(1, activation = 'sigmoid')(hidden2)
```

```python
model = Model(inputs = input, outputs = [output1, output2])
```

```python
from keras.utils import plot_model
plot_model(model, show_shapes = True)
```

# **Multiple Input Model**

```python
inputA = Input(shape = (128,))
inputB = Input(shape = (32,))
```

```python
A1= Dense(64, activation = 'relu')(inputA)
A2= Dense(32, activation = 'relu')(A1)
A3= Dense(4, activation= 'relu')(A2)

B1= Dense(8, activation = 'relu')(inputB)
B2= Dense(4, activation = 'relu')(B1)
```

```python
combined = Concatenate()([A3, B2])
```

```python
C1 = Dense(2, activation = 'relu')(combined)
C2 = Dense(1, activation = 'sigmoid')(C1)
```

```python
model2 = Model(inputs = [inputA, inputB], outputs = C2)
```

```python
model2.summary()
```

```python
from keras.utils import plot_model
plot_model(model2)
```
