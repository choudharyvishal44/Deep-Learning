# Exponentially Weighted Moving Average


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv(r"C:\Users\Dell\Downloads\archive (34)\climate_data.csv", encoding = 'latin-1')
```

```python
df = df[['date','RR']].iloc[0:100]
```

```python
plt.scatter(x = 'date', y = 'RR', data = df)
plt.show()
```

### Applying EWM

```python
x1 = df['RR'].ewm(alpha = 0.9).mean()
print(x1)
```

```python
plt.scatter(x = 'date', y = 'RR',data = df, label = 'Rainfall')
plt.plot(x1, color = 'red', label = 'EWM')
plt.legend()
plt.show()
```
